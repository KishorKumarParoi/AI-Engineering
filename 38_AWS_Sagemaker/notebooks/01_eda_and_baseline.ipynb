# 📱 Mobile Price Classification: EDA & Baseline Model

This notebook explores the Mobile Price Classification dataset, evaluates feature correlations,
and trains an initial Scikit-Learn baseline classifier before moving into AWS SageMaker Pipelines.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score

# Load dataset
df = pd.read_csv("../data/raw/mob_price_classification_train.csv")
print("Dataset Shape:", df.shape)
df.head()
```

## 1. Target Class Distribution
Let us verify the balance of our multi-class target:  (0=Low, 1=Medium, 2=High, 3=Very High).

```python
class_counts = df["price_range"].value_counts().sort_index()
print("Target Class Distribution:
", class_counts)
print("Null counts across features:
", df.isnull().sum().sum())
```

## 2. Feature Correlation with Target
RAM, Battery Power, and Pixel Resolution are typically top predictors for mobile price tier.

```python
correlations = df.corr()["price_range"].sort_values(ascending=False)
print("Top Positive Correlations with price_range:
", correlations.head(6))
print("
Top Negative Correlations with price_range:
", correlations.tail(3))
```

## 3. Train Baseline Scikit-Learn Random Forest

```python
X = df.drop(columns=["price_range"])
y = df["price_range"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.15, random_state=42, stratify=y)

baseline_rf = RandomForestClassifier(n_estimators=100, random_state=42)
baseline_rf.fit(X_train, y_train)

preds = baseline_rf.predict(X_test)
print("Holdout Test Accuracy:", accuracy_score(y_test, preds))
print("
Classification Report:
", classification_report(y_test, preds))
```
